```sql
CREATE OR REPLACE TABLE ecommerce_analytics.data_warehouse.fact_orders_item(
  order_id BIGINT,
  customer_id BIGINT,
  product_id STRING,
  promo_id INT,
  order_date_key INT,
  quantity INT,
  price FLOAT,
  amount FLOAT
);

INSERT INTO ecommerce_analytics.data_warehouse.fact_orders_item
SELECT 
  order_number AS order_id,
  customer_id,
  product_id,
  promo_id,
  date_format(to_timestamp(order_datetime),'yyyyMMdd') AS order_date_key,
  qty AS quantity,
  price,
  qty * price AS amount
FROM batch_1.data1.sales_order;

SELECT * FROM ecommerce_analytics.data_warehouse.fact_orders_item LIMIT 10;
```
